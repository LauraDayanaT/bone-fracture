```python
import torch
from torch import nn, optim
import torch.nn.functional as F
from torchvision import datasets, transforms
from torch.utils.data import DataLoader, TensorDataset
```

```python
class RedNeuronal(nn.Module):
    def __init__(self, entrada, ocultas, salida):
        super().__init__()
        self.fc1 = nn.Linear(entrada, ocultas[0])
        self.fc2 = nn.Linear(ocultas[0], ocultas[1])
        self.salida = nn.Linear(ocultas[1], salida)
        
    def forward(self, x):
        x = x.view(x.shape[0], -1)
        x = F.relu(self.fc1(x))
        x = F.relu(self.fc2(x))
        x = F.log_softmax(self.salida(x), dim=1)
        return x
```

```python
# Datos de prueba para evitar bloqueos en la descarga
x_mnist = torch.randn(1000, 1, 28, 28)
y_mnist = torch.randint(0, 10, (1000,))

trainset_mnist = TensorDataset(x_mnist, y_mnist)
trainloader_mnist = DataLoader(trainset_mnist, batch_size=64, shuffle=True)
print("¡MNIST listo para entrenar!")
```

```python
modelo_mnist = RedNeuronal(entrada=784, ocultas=[128, 64], salida=10)
criterio = nn.NLLLoss()
optimizador = optim.Adam(modelo_mnist.parameters(), lr=0.003)

epocas = 5
for e in range(epocas):
    perdida_acumulada = 0
    for imagenes, etiquetas in trainloader_mnist:
        optimizador.zero_grad()
        salida = modelo_mnist(imagenes)
        perdida = criterio(salida, etiquetas)
        perdida.backward()
        optimizador.step()
        perdida_acumulada += perdida.item()
    print(f"Época {e+1}/{epocas} - Pérdida MNIST: {perdida_acumulada/len(trainloader_mnist):.4f}")
```

```python
# Generación de datos de radiografías simulados (1000 imágenes de 1x128x128)
x_fracturas = torch.randn(1000, 1, 128, 128)
y_fracturas = torch.randint(0, 2, (1000,)) # 2 clases: 0 (No Fractura), 1 (Fractura)

dataset_fracturas = TensorDataset(x_fracturas, y_fracturas)
loader_fracturas = DataLoader(dataset_fracturas, batch_size=32, shuffle=True)
print("¡Bone Fracture listo para entrenar!")
```

```python
modelo_fracturas = RedNeuronal(entrada=128*128, ocultas=[256, 64], salida=2)
criterio_fracturas = nn.NLLLoss()
optimizador_fracturas = optim.Adam(modelo_fracturas.parameters(), lr=0.001)

epocas = 5
for e in range(epocas):
    perdida_acumulada = 0
    for imagenes, etiquetas in loader_fracturas:
        optimizador_fracturas.zero_grad()
        salida = modelo_fracturas(imagenes)
        perdida = criterio_fracturas(salida, etiquetas)
        perdida.backward()
        optimizador_fracturas.step()
        perdida_acumulada += perdida.item()
    print(f"Época {e+1}/{epocas} - Pérdida Bone Fracture: {perdida_acumulada/len(loader_fracturas):.4f}")
```
