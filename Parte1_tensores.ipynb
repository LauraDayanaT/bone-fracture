```python
import torch

print("Versión de PyTorch:", torch.__version__)

# 1. Un tensor 1D → como una fila de casillas
x = torch.tensor([1, 2, 3])
print("Vector:", x, "| shape:", x.shape)

# 2. Una radiografía en blanco y negro → tensor 2D (alto x ancho de píxeles)
radiografia = torch.rand(5, 5)  # 5x5 píxeles con valores random entre 0 y 1
print("Radiografía simulada:\n", radiografia)
```

```python
# Suma y multiplicación entre tensores
a = torch.rand(3, 3)
b = torch.rand(3, 3)

suma = a + b
producto = a * b          # multiplicación elemento por elemento
producto_matricial = a @ b  # multiplicación matricial real (como en álgebra lineal)

print("Suma:\n", suma)
print("Producto elemento a elemento:\n", producto)
print("Producto matricial:\n", producto_matricial)

# Reshape: cambiar la forma sin cambiar los datos
x = torch.rand(6)
print("Original:", x.shape)
print("Reacomodado a 2x3:", x.reshape(2, 3).shape)
```

```python
x = torch.rand(3, 2)
y = torch.ones(x.size())
z = x + y

print(z)
print(z[0])       # primera fila
print(z[:, 1:])   # todas las filas, desde la columna 1 en adelante

# Operación normal: crea un tensor NUEVO, no toca el original
z.add(1)
print("z sigue igual:\n", z)

# Operación in-place (con guion bajo _): SÍ modifica z directamente
z.add_(1)
print("z ya cambió:\n", z)
```

```python
import numpy as np

a = np.random.rand(4, 3)
b = torch.from_numpy(a)   # de numpy a tensor
print(b)

de_vuelta = b.numpy()     # de tensor a numpy
print(de_vuelta)

#comparten la misma memoria
b.mul_(2)   # multiplico el tensor por 2, in-place
print(a)    # array numpy cambio
```
