```python
import torch
from torch import nn
import torch.nn.functional as F
```

```python
class RedSimple(nn.Module):
    def __init__(self, entrada, ocultas, salida):
        super().__init__()
        self.fc1 = nn.Linear(entrada, ocultas[0])
        self.fc2 = nn.Linear(ocultas[0], ocultas[1])
        self.salida = nn.Linear(ocultas[1], salida)
        
    def forward(self, x):
        x = x.view(x.shape[0], -1)  # Aplanar la imagen
        x = F.relu(self.fc1(x))
        x = F.relu(self.fc2(x))
        x = F.log_softmax(self.salida(x), dim=1)
        return x
```

```python
# MNIST: Imágenes 28x28 (784 píxeles), 10 clases (0-9)
modelo_mnist = RedSimple(entrada=784, ocultas=[128, 64], salida=10)

# Simulación de un lote de 64 imágenes
imagenes_mnist = torch.randn(64, 1, 28, 28)
probabilidades_mnist = torch.exp(modelo_mnist(imagenes_mnist))

print("Salida MNIST:", probabilidades_mnist.shape)
```

```python
# Bone Fracture: Imágenes 128x128 (16384 píxeles), 2 clases (Fractura / No Fractura)
modelo_fracturas = RedSimple(entrada=128*128, ocultas=[256, 64], salida=2)

# Simulación de un lote de 32 radiografías
imagenes_fractura = torch.randn(32, 1, 128, 128)
probabilidades_fractura = torch.exp(modelo_fracturas(imagenes_fractura))

print("Salida Fracturas:", probabilidades_fractura.shape)
```
